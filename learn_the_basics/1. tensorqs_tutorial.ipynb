Tensors 张量
=======

Tensors are a specialized data structure that are very similar to arrays
and matrices. In PyTorch, we use tensors to encode the inputs and
outputs of a model, as well as the model's parameters.

张量（Tensor）是一种特殊的数据结构，与数组和矩阵非常相似。在 PyTorch 中，我们用张量来表示模型的输入和输出，以及模型的参数。

Tensors are similar to [NumPy's](https://numpy.org/) ndarrays, except
that tensors can run on GPUs or other hardware accelerators. In fact,
tensors and NumPy arrays can often share the same underlying memory,
eliminating the need to copy data (see
`bridge-to-np-label`{.interpreted-text role="ref"}). Tensors are also
optimized for automatic differentiation (we\'ll see more about that
later in the [Autograd](autogradqs_tutorial.html) section). If you're
familiar with ndarrays, you'll be right at home with the Tensor API. If
not, follow along!

张量类似于 [NumPy](https://numpy.org/) 的 ndarray，不同之处在于张量可以在 GPU 或其他硬件加速器上运行。事实上，张量和 NumPy 数组通常可以共享同一块底层内存，无需复制数据（参见下文「与 NumPy 的桥接」一节）。张量还针对自动微分做了优化（我们会在后面的 [Autograd](autogradqs_tutorial.html) 一节中详细介绍）。如果你熟悉 ndarray，那么 Tensor API 对你来说会很容易上手；如果不熟悉，也跟着学下去吧！


```python
import torch
import numpy as np
```

Initializing a Tensor 初始化张量
=====================

Tensors can be initialized in various ways. Take a look at the following
examples:

张量可以通过多种方式初始化。请看下面的例子：

**Directly from data 直接从数据创建**

Tensors can be created directly from data. The data type is
automatically inferred.

张量可以直接从数据创建，数据类型会被自动推断。


```python
data = [[1, 2], [3, 4]]
x_data = torch.tensor(data)

print(x_data)
```

**From a NumPy array 从 NumPy 数组创建**

Tensors can be created from NumPy arrays (and vice versa - see
`bridge-to-np-label`{.interpreted-text role="ref"}).

张量可以从 NumPy 数组创建（反之亦然，参见「与 NumPy 的桥接」一节）。


```python
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
print(x_np)
```

**From another tensor: 从另一个张量创建：**

The new tensor retains the properties (shape, datatype) of the argument
tensor, unless explicitly overridden.

除非显式覆盖，否则新张量会保留参数张量的属性（形状、数据类型）。


```python
x_ones = torch.ones_like(x_data)
print(f"Ones Tensor: \n {x_ones} \n")
print(torch.ones(1, 2))

x_rand = torch.rand_like(x_data, dtype=torch.float)
print(f"Random Tensor: \n {x_rand} \n")

```

**With random or constant values: 使用随机值或常数值：**

`shape` is a tuple of tensor dimensions. In the functions below, it
determines the dimensionality of the output tensor.

`shape` 是一个表示张量维度的元组。在下面的函数中，它决定了输出张量的维度。


```python
shape = (2, 3)
rand_tensor = torch.rand(shape)
randn_tensor = torch.randn(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(rand_tensor)
print(randn_tensor)
print(ones_tensor)
print(zeros_tensor)
```

Attributes of a Tensor 张量的属性
======================

Tensor attributes describe their shape, datatype, and the device on
which they are stored.

张量的属性描述了它的形状、数据类型以及存储它的设备。


```python
tensor = torch.rand(3, 4)
print(tensor)
print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```

Operations on Tensors 张量运算
=====================

Over 1200 tensor operations, including arithmetic, linear algebra,
matrix manipulation (transposing, indexing, slicing), sampling and more
are comprehensively described
[here](https://pytorch.org/docs/stable/torch.html).

[这里](https://pytorch.org/docs/stable/torch.html)全面介绍了 1200 多种张量运算，包括算术运算、线性代数、矩阵操作（转置、索引、切片）、采样等。

Each of these operations can be run on the CPU and
[Accelerator](https://pytorch.org/docs/stable/torch.html#accelerators)
such as CUDA, MPS, MTIA, or XPU. If you're using Colab, allocate an
accelerator by going to Runtime \> Change runtime type \> GPU.

这些运算都可以在 CPU 上运行，也可以在 CUDA、MPS、MTIA 或 XPU 等[加速器](https://pytorch.org/docs/stable/torch.html#accelerators)上运行。如果你使用 Colab，可以通过 Runtime \> Change runtime type \> GPU 来分配加速器。

By default, tensors are created on the CPU. We need to explicitly move
tensors to the accelerator using `.to` method (after checking for
accelerator availability). Keep in mind that copying large tensors
across devices can be expensive in terms of time and memory!

默认情况下，张量创建在 CPU 上。我们需要使用 `.to` 方法（先检查加速器是否可用）把张量显式移动到加速器上。请记住，在设备之间复制大型张量，在时间和内存上的开销都可能很大！


```python
# We move our tensor to the current accelerator if available

if torch.accelerator.is_available():
    tensor = tensor.to(torch.accelerator.current_accelerator())
```

Try out some of the operations from the list. If you\'re familiar with
the NumPy API, you\'ll find the Tensor API a breeze to use.

试试列表中的一些运算。如果你熟悉 NumPy API，会发现 Tensor API 用起来非常轻松。


**Standard numpy-like indexing and slicing: 类似 NumPy 的标准索引和切片：**


```python
tensor = torch.ones(4, 4)
print(f"First row: {tensor[0]}")
print(f"First column: {tensor[:, 0]}")
print(f"Last column: {tensor[..., -1]}")
tensor[:, 1] = 0
print(tensor)
```

**Joining tensors** You can use `torch.cat` to concatenate a sequence of
tensors along a given dimension. See also
[torch.stack](https://pytorch.org/docs/stable/generated/torch.stack.html),
another tensor joining operator that is subtly different from
`torch.cat`.

**拼接张量** 你可以使用 `torch.cat` 沿指定维度拼接一系列张量。另请参阅 [torch.stack](https://pytorch.org/docs/stable/generated/torch.stack.html)，它是另一个拼接张量的运算，与 `torch.cat` 有细微的差别。


```python
t1 = torch.cat([tensor, tensor, tensor], dim=0)
print(t1)

t2 = torch.cat([tensor, tensor, tensor], dim=1)
print(t2)

```

**Arithmetic operations 算术运算**


```python
# This computes the matrix multiplication between two tensors. y1, y2, y3 will have the same value
# ''tensor.T'' returns the transpose of a tensor
y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)

y3 = torch.rand_like(y1)
print(y3)
torch.matmul(tensor, tensor.T, out = y3)
print(y1, y2, y3)




# This computes the element-wise product. z1, z2, z3 will have the same value
z1 = tensor * tensor
z2 = tensor.mul(tensor)

z3 = torch.rand_like(tensor)
print(z3)
torch.mul(tensor, tensor, out=z3)
print(z1, z2, z3)
```

**Single-element tensors** If you have a one-element tensor, for example
by aggregating all values of a tensor into one value, you can convert it
to a Python numerical value using `item()`:

**单元素张量** 如果你有一个只含一个元素的张量（例如把张量的所有值聚合成一个值），可以使用 `item()` 把它转换为 Python 数值：


```python
agg = tensor.sum()
agg_item = agg.item()
print(tensor)
tensor2 = tensor.type(torch.int64)
print(tensor2)
print(agg)
print(agg_item, type(agg_item))
```

**In-place operations** Operations that store the result into the
operand are called in-place. They are denoted by a `_` suffix. For
example: `x.copy_(y)`, `x.t_()`, will change `x`.

**原地操作** 把结果存回操作数中的运算称为原地（in-place）操作，它们以 `_` 后缀表示。例如：`x.copy_(y)`、`x.t_()` 都会改变 `x`。


```python
# 广播机制

print(f"{tensor} \n")
tensor.add_(5)
print(tensor)
```

Bridge with NumPy 与 NumPy 的桥接 {#bridge-to-np-label}
=================

Tensors on the CPU and NumPy arrays can share their underlying memory
locations, and changing one will change the other.

CPU 上的张量和 NumPy 数组可以共享底层内存位置，改变其中一个，另一个也会随之改变。


Tensor to NumPy array 张量转 NumPy 数组
=====================


```python
t = torch.ones(5)
print(f"t: {t}")
n = t.numpy()
print(f"n: {n}")

```

A change in the tensor reflects in the NumPy array.

张量中的变化会反映到 NumPy 数组中。


```python
t.add_(1)
print(f"t: {t}")
print(f"n: {n}")
```

NumPy array to Tensor NumPy 数组转张量
=====================


```python
# from_numpy使两者共享内存
n = np.ones(5)
t = torch.from_numpy(n)
print(n)
print(t)
```

Changes in the NumPy array reflects in the tensor.

NumPy 数组中的变化会反映到张量中。


```python
np.add(n, 1, out=n)
print(f"t: {t}")
print(f"n: {n}")
```
